#What Is a Tool in LangChain?

**Definition**  

A Tool is a mechanism that gives an LLM the ability to perform external tasks by connecting it with functions.

In simple words:

**Tools provide hands and legs to an LLM so that it can perform tasks instead of only thinking and speaking.**

**--Technical Definition**  

A tool is a function containing the logic required to perform a particular task. This function is packaged in a way that allows an LLM to understand and call it when needed.

**--Working Flow of Tools**  

User Query → LLM Understands the Query → LLM Selects the Required Tool → Tool Executes the Task → Result Returns to LLM → LLM Responds to User.

**--Types of Tools in LangChain**   

LangChain provides two major types of tools:

**1. Built-in Tools**

Built-in tools are already created and provided by the LangChain team for commonly required tasks.
      
                                       **OR**

A built-in tool is a tool that LangChain already provides for you —it’s pre-built, productionready, and requires minimal or no setup.

You don’t have to write the function logic yourself — you just import and use it.

**Examples:**

1.Google Search

2.Internet Search

3.Wikipedia Search

4.Command-line operations

**Why do we need built-in tools?**

Many applications require common functionalities. Instead of writing the entire code ourselves, we can directly use tools provided by LangChain.

**Advantages:**

1.Ready to use.

2.No need to write the complete functionality from scratch.

3.Useful for common tasks.

4.Saves development time.

# Custom Tools :
Custom tools are tools that developers create according to their own requirements.

**--Why do we need custom tools?**

Sometimes an application has specific requirements that cannot be fulfilled by the available built-in tools.

**For example :-** a company may need a tool that:

Fetches its internal employee information.

Checks company-specific records.

Performs a particular business operation.

In such cases, developers can create their own custom tools.

**--Advantages:**

1.Designed for specific requirements.

2.Can implement company-specific functionality.

3.Gives developers more control over task execution.


**-->3 Ways to Create Custom Tools in LangChain :**   
In LangChain, we can create custom tools in three main ways:

1) Using the **@tool decorator**

2) Using **StructuredTool with Pydantic**

3) Using the **BaseTool class**

```python
from langchain_community.tools import tool
```

# Method 1 -  Using the @tool Decorator :  

--This is the simplest and most commonly used method to create custom tools in LangChain.

--We create a normal Python function and use the @tool decorator to convert it into a LangChain tool.  
**--How does it work?**
There are three steps:
1. **Create a function:** Write the logic for the task you want to perform.

2. **Add type hints and a docstring:** Type hints describe the input and output types, while the docstring explains what the function does.

3. **Apply the @tool decorator:** This converts the function into a LangChain tool.

```python
# Step 1 - Create a function

def multiply(a,b):
  """Multiply two numbers""" # docstring that tell that the function should multiply the two numbers.
  return a*b
```

```python
# Step 2 - add type hints

def multiply(a: int, b:int) -> int: # Type Hints tells here that input(a,b) should be int type and output(-> int) should be type int.
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 - Add tool decorator

@tool
def multiply(a:int, b:int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
result = multiply.invoke({"a":3, "b":5})
```

```python

result
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

# Method 2 - Using StructuredTool :
**A StructuredTool in LangChain is a special type of tool where the input follows a structured schema, typically defined using a Pydantic model.**

In simple words:

StructuredTool allows us to define and validate the inputs of our tool more strictly.

In the first method, we used Python type hints to describe the inputs. But with Pydantic, we can also define additional rules, such as minimum values, maximum values, and required fields.

                                 OR

**StructuredTool:**
- A special type of tool in LangChain.
- It allows a tool to accept multiple structured inputs.
- Its input follows a defined schema.

**Pydantic:**
- Used to define and validate the input schema.
- Specifies the input field names, data types, and validation rules.   
**For example :** a weather tool might accept:
   - city: string
   - country: string  

This ensures that the tool receives inputs in the expected structure.

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
# Create a Pydantic model OR args schema using pydantic
class MultiplyInput(BaseModel):  # MultiplyInput defines the input structure of our tool.
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
# Create a function

def multiply_func(a:int,b:int)->int:
  """Multiply two numbers"""
  return a*b
```

```python

multiply_tool = StructuredTool.from_function(
    func=multiply_func,     # func ---> Specifies the Python function.
    name="multiply",        # name ---> Defines the tool name.
    description="Multiply two numbers",  # description ---> Explains what the tool does OR purpose of tool.
    args_schema=MultiplyInput            # args_schema ---> Defines the input schema using Pydantic.
)
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Method 3 - Using BaseTool Class :  
BaseTool is the abstract base class for all tools in LangChain.

It defines the core structure and interface that every tool must follow, whether it is a simple tool or a fully customized tool.

Other tool types, such as @tool and StructuredTool, are built on top of BaseTool.

**--What does this mean?**   

Think of BaseTool as a parent class.

When we create a custom tool by inheriting from BaseTool, we can define its behavior and customize how it works.

This method provides more control over the tool's implementation.

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# Step 1: Define input schema
# Create a Pydantic model OR args schema using pydantic

class MultiplyInput(BaseModel):  # MultiplyInput defines the input structure of our tool.
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
# Step 2: Create custom tool class
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
# Step 3: Create tool object
multiply_tool = MultiplyTool()
```

```python
# Step 4: Execute tool
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

# Toolkit :
A Toolkit in LangChain is a collection of related tools grouped together to perform specific types of tasks.

In simple English:

**- Tool:** A single function that performs one specific task.
  
**- Toolkit:** A collection of tools that work together to perform multiple related tasks.  

**Real-world example :**
Suppose you are building an AI agent that works with a SQL database.

Your agent needs to perform multiple tasks:
1. List all available tables in the database.
2. Check the structure of a particular table.
3. Execute SQL queries.
4. Check whether an SQL query is correct.

Instead of creating and managing all these tools separately, you can use a SQLDatabaseToolkit, which provides related database tools together.

**Advantages :**

1) **Reusability:** Use the same collection of tools across multiple applications.

2) **Organization:** Keep related tools together.

3) Easy Management: Manage a group of related tools more conveniently.

4) **Modularity:** Separate different functionalities into reusable

```python
from langchain_core.tools import tool

# Step 1: Create related tools

@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
# Step 2: Create a custom toolkit

class MathToolkit:
  def get_tools(self):
    return [add,multiply]
```

```python
# Step 3: Create toolkit object

toolkit = MathToolkit()

# Step 4: Get all tools

tools = toolkit.get_tools()

# Step 5: Access tools

for tool in tools:
  print(tool.name, "=>", tool.description)
```

```python
# Step 6: Execute tools

print("Addition of Two Number :",add.invoke({"a": 10, "b": 5}))
print("Multiplication of two number :",multiply.invoke({"a": 10, "b": 5}))
```
